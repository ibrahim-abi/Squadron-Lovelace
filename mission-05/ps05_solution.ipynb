## Part 1: Systems Architect — HNSW Multi-Layer Vector Proximity Index
Constructs multi-layer proximity graphs with logarithmic skip-list traversal (Malkov & Yashunin, 2018).

```python
from importlib import import_module
import numpy as np
hnsw_mod = import_module("01_HNSW_Vector_Index")
HierarchicalNSW = hnsw_mod.HierarchicalNSW

np.random.seed(42)
dim = 64
hnsw = HierarchicalNSW(dim=dim, M=8, ef_construction=32)
for i in range(200):
    hnsw.insert(i, np.random.randn(dim).astype(np.float32))

q = np.random.randn(dim).astype(np.float32)
matches = hnsw.query(q, top_k=5)
print("=== Subsystem 05: HNSW Vector Proximity Index Simulation ===")
print("Top 5 Nearest Astrogation Neighbors (ID, Similarity):")
for doc_id, sim in matches:
    print(f"  Node {doc_id:04d} -> Cosine Similarity: {sim:.4f}")
print("HNSW Skip-Graph Traversal Verified!")
```

## Part 2: Communications & Protocol Officer — Okapi BM25 & Reciprocal Rank Fusion (RRF)
Exact keyword preservation and dual-path rank aggregation (k=60).

```python
bm25_mod = import_module("02_BM25_and_RRF_Combiner")
BM25Okapi = bm25_mod.BM25Okapi
reciprocal_rank_fusion = bm25_mod.reciprocal_rank_fusion

corpus = [
    "AURA Subsystem 05 contains the dual-path associative memory core in Sector SEC-05-M01.",
    "Pulsar radiation in Crab nebula induces high frequency electromagnetic interference at 30.22 Hz.",
    "Warp vector plasma manifold alignment code PR-882-HEX must be phase locked.",
    "Emergency magnetic deflector shield harmonic code is MAG-SHIELD-77B.",
    "Cryo-coolant emergency purge valve serial number is VLV-904-CRY.",
    "General astrogation protocol: ungrounded parametric models suffer from severe confabulation."
]
bm25 = BM25Okapi(corpus)
sparse_scores = bm25.score("PR-882-HEX plasma manifold")
print("=== Subsystem 05: BM25 Lexical Retrieval Demonstration ===")
for doc_id, sc in sparse_scores[:3]:
    print(f"Doc {doc_id} [BM25: {sc:.4f}] -> {corpus[doc_id]}")

dense_ranks = [0, 2, 5, 1, 3, 4]
sparse_ranks = [d for d, _ in sparse_scores]
fused = reciprocal_rank_fusion(dense_ranks, sparse_ranks, k=60)
print("=== Subsystem 05: Reciprocal Rank Fusion (k=60) Hybrid Search ===")
for doc_id, rrf_val in fused[:2]:
    print(f"Fused Rank -> Doc {doc_id} (RRF Score: {rrf_val:.6f}) => {corpus[doc_id]}")
print("Hybrid Retrieval Verified!")
```
